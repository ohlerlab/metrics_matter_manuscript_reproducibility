```python
def plot_umaps(models):
    np.random.seed(61)

    # Collect computed scores, nested dict is simple to convert to pd.DataFrame
    for model in models:
        # Load data
        embedding = ad.AnnData(X=pd.read_csv('./../../data/simulated_2d/{}.csv'.format(model))[['X', 'Y']].values, obs=pd.read_csv('./../../data/simulated_2d/{}.csv'.format(model))[['sample', 'cell_type']])
        embedding.obs['Batch'] = embedding.obs['sample'].astype('category')
        embedding.obs['Cell_type'] = embedding.obs['cell_type'].astype('category')
        
        # assign colors
        embedding.uns['Batch_colors'] = ['#882255', '#CC6677', '#332288', '#88CCEE']
        #embedding.uns['cell_type_colors'] = ['#DDAA33', '#117733', '#EE7733']
        embedding.uns['Cell_type_colors'] = ['#FEC44F']

        embedding.obsm['Dim'] = embedding.X

        # random plotting order
        embedding = sc.pp.subsample(embedding, fraction=1., copy=True)

        ### Save UMAP visualization as png 
        ax = sc.pl.embedding(
            adata=embedding,
            basis='Dim',
            color=['Batch'],
            size=3,
            #frameon=False,
            ncols=1,
            wspace=-0.2,
            #hspace=0.8,
            #save="_{}_sim_simplified_nested_single_cell_type.png".format(model),
            legend_loc=None,
            title=[''], 
            show=False,
            #legend_fontsize='x-small'
        )

        ax.set_xlim(-10, 10)  # Replace with your desired x-axis limits
        ax.set_ylim(-10, 10)  # Replace with your desired y-axis limits
        plt.savefig('./../../figures/{}.png'.format(model), bbox_inches='tight')
        plt.show()

        
    ### Save UMAP visualization as png 
    ax = sc.pl.embedding(
            adata=embedding,
            basis='Dim',
            color=['Batch'],
            size=3,
            #frameon=False,
            ncols=1,
            wspace=-0.2,
            #hspace=0.8,
            #save="_{}_sim_simplified_nested_single_cell_type_legend.png".format(model),
            legend_fontsize='x-small',
            title=[''],
            show=False,
    )
    
    ax.set_xlim(-10, 10)  # Replace with your desired x-axis limits
    ax.set_ylim(-10, 10)  # Replace with your desired y-axis limits
    plt.savefig('./../../figures/{}_legend.png'.format(model), bbox_inches='tight')
    plt.show()
```
